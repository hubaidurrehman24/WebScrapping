## **Task 1: Static Website - Extract Information About the CodeTechMentor Creator**

### **URL:**  
[https://www.codetechmentor.com/p/about-me.html](https://www.codetechmentor.com/p/about-me.html)

### **Objective:**  
Write a Python script that uses the `requests` library to fetch the HTML content of the CodeTechMentor "About" page. Use `BeautifulSoup` to parse the HTML and extract the following details about the creator:

### **Requirements:**
1. **Creator's Name**: Extract the name of the CodeTechMentor creator.
2. **Introduction Text**: Retrieve and combine all the introductory text.
3. **Social Media Links**: Find social media links.

### **Expected Output Format:**
- Print the extracted details in a structured format for easy readability.
- Ensure the script handles cases where elements might be missing.
- Provide appropriate messages if any of the required information is not found.


```python
# Start Code from Here...
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
URL="https://www.codetechmentor.com/p/about-me.html"
page=requests.get(URL)
print(page)
```

```python
soup = BeautifulSoup(page.content,"html.parser")
# 1- Creator's name
name= soup.find('span',{'class':'highlight'}).get_text()

print("Cretaer's name : ",  name )
# 2- Text Extraction
introduction_divs=soup.find_all('div',class_='intro')
introduction_text=[]
for t in introduction_divs:
    para=t.find_all('p')
    for p in para:
        introduction_text.append(p.get_text().strip())
print("Introduction Text : ")
for intro_text in introduction_text:
    print(intro_text)

# 3- social links
social_icons=soup.find_all('div',class_='social-icons')
social_links=[]
for s in social_icons:
    anchor=s.find_all('a')
    for a in anchor:
        link=a['href']
        social_links.append(link)
        
print("Social Links : ")
for links in social_links:
    print(links)
                        

                
```

## **Task 2: Static Website Data Extraction (PriceOye Samsung Tablets)**

### **URL:**  
[https://priceoye.pk/tablets/samsung](https://priceoye.pk/tablets/samsung)

### **Objective:**  
Write a Python script that uses the `requests` library to retrieve the HTML content of the PriceOye Samsung tablets webpage. Use `BeautifulSoup` to parse the HTML and extract the following details:

### **Requirements:**
1. **Tablet Names**: Extract the names of all Samsung tablets.
2. **Prices**: Retrieve the listed prices of the tablets.
3. **Discounts**: Extract any discount information available.
4. **Exclusions**: Ensure that any products marked as **discontinued** are excluded.

### **Expected Output Format:**
- Print the extracted details in a structured format for easy readability.

```python
# Start Code from Here...
URL="https://priceoye.pk/tablets/samsung"
page=requests.get(URL)
soup=BeautifulSoup(page.content,"html.parser")

productbox=soup.find_all('div',class_='productBox')
productsList=[]

for product in productbox:
    if "DISCONTINUED" not in product.text:
        name = product.find('div',class_='p-title')
        tablet_name=name.get_text().strip()if name else "Found No Name"

        price=product.find('div',class_='price-box')
        tablet_price=price.get_text().strip()if price else "Found no price"

        discount=product.find('div',class_='price-diff-saving')
        tablet_discount = discount.get_text().strip() if discount else 'No disocunt'
        productsList.append({
            'Tablet Name' : tablet_name,
            'Price'  :  tablet_price,
            'Discount' : tablet_discount
        })
for product in productsList:
     print(f"Name : {product['Tablet Name']}")
     print(f"Price : {product['Price']}") 
     print(f"Discount : {product['Discount']}")   
     print("-"*30)   
```

```python


first_link=driver.find_element(By.XPATH,"(//h3)[1]")
first_link.click()
wait=WebDriverWait(driver,20)

 
table = driver.find_element(By.CLASS_NAME,"table").find_element(By.TAG_NAME,"tbody")
rows = table.find_elements(By.TAG_NAME, "tr")
three_months_ago = datetime.today() - timedelta(days=90)    
data = []
for row in rows:
    cols = row.find_elements(By.TAG_NAME, "td")
    if len(cols) == 7: 
            date = cols[0].text
            date_str = datetime.strptime(date, "%b %d, %Y")
            if date_str >= three_months_ago:
                 
               open_price = cols[1].text
               high = cols[2].text
               low = cols[3].text
               close = cols[4].text
               adj_close = cols[5].text
               volume = cols[6].text
               data.append([date_str, open_price, high, low, close, adj_close, volume])
    
    
    
    df = pd.DataFrame(data, columns=["Date", "Open", "High", "Low", "Close", "Adj Close", "Volume"])
    
print(df)
```

## **Task 5: Dynamic Website Data Extraction (YouTube Video Content - Aima Baig)**

### **Objective:**  
Use **Selenium** and **BeautifulSoup** to extract **video titles, view counts, and comment counts** from **Aima Baig's official YouTube channel**.

---

### **Steps for Implementation:**

### **Step 1: Navigate to the YouTube Channel**
- Use Selenium to open the **YouTube URL:**  
  🔗 [Aima Baig's YouTube Channel](https://www.youtube.com/@RealAimaBaig/videos)
- Ensure the page fully **loads** before extracting data.

---

### **Step 2: Extract Video Titles and View Counts**
- Use **BeautifulSoup** to parse the page content.
- Locate the section containing the **list of videos**.
- Extract the **titles** of each video and their corresponding **view counts**.

---

### **Step 3: Navigate to Each Video Page for Comments Count**
- For each video, use **Selenium** to click on the **video link**, opening it in a new tab.
- Since the **comments section** loads dynamically as you scroll, use **Selenium’s scrolling method** to scroll down until comments are visible.

---

### **Step 4: Extract the Comments Count**
- Once the comments section is loaded, use **BeautifulSoup** or **Selenium** to locate and extract the **number of comments** for each video.

---

### **Step 5: Print the Results**
- After collecting all the data (**video titles, views, comments count**), **print** the results in a well-structured format.

```python
# Start Code from Here...
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
from selenium.common.exceptions import TimeoutException, NoSuchElementException
```

```python
options = webdriver.ChromeOptions()
# options.add_argument('--headless')  # Run in headless mode
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
# options.binary_location = '/usr/bin/google-chrome'

# Set up ChromeDriver service
service = Service(ChromeDriverManager().install())

# Initialize the WebDriver with Chrome
driver = webdriver.Chrome(service=service, options=options)
wait = WebDriverWait(driver, 10)

# Open the YouTube page
driver.get("https://www.youtube.com/@RealAimaBaig/videos")
print("Title:", driver.title)

# Wait for the videos to load
wait.until(EC.presence_of_element_located((By.TAG_NAME, "ytd-rich-grid-media")))
soup = BeautifulSoup(driver.page_source, "html.parser")
video_elements = driver.find_elements(By.CSS_SELECTOR, 'ytd-rich-grid-media')

# Extract video titles and view counts
videos = soup.find_all('ytd-rich-grid-media')

for index, video_element in enumerate(video_elements):

    title_element = videos[index].find('yt-formatted-string', {'id': 'video-title'})
    view_count_element = videos[index].find('span', {'class': 'inline-metadata-item'})

    if title_element and view_count_element:
        title = title_element.text.strip()
        view_count = view_count_element.text.strip()
        print(f"Title: {title}")
        print(f"Views: {view_count}")

    # Click the video to open it
    video_element.find_element(By.ID, 'video-title').click()

    # Wait for the video page to load
    wait.until(EC.presence_of_element_located((By.TAG_NAME, "ytd-comments")))

    # Scroll down to load comments
    last_height = driver.execute_script("return document.documentElement.scrollHeight")
    while True:
        driver.execute_script("window.scrollBy(0, 2000);") #scroll down
        time.sleep(2)
        new_height = driver.execute_script("return document.documentElement.scrollHeight")
        if new_height == last_height:
            break
        last_height = new_height

    try:
        comment_count_element = driver.find_element(By.CSS_SELECTOR, "#count .count-text")
        comment_count = comment_count_element.text
        print(f"Comments: {comment_count}\n")
    except:
        print("Comments: Unable to find comment count.\n")

    driver.back()
    time.sleep(2)

driver.quit()
```
